```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.ml.feature import VectorAssembler, StandardScaler
from pyspark.ml.clustering import KMeans
import matplotlib.pyplot as plt
```

```python
base_path = "/Volumes/workspace/default/olist_volume/"

customers = spark.read.option("header", True).option("inferSchema", True) \
    .csv(base_path + "olist_customers_dataset.csv")

orders = spark.read.option("header", True).option("inferSchema", True) \
    .csv(base_path + "olist_orders_dataset.csv")

order_items = spark.read.option("header", True).option("inferSchema", True) \
    .csv(base_path + "olist_order_items_dataset.csv")

payments = spark.read.option("header", True).option("inferSchema", True) \
    .csv(base_path + "olist_order_payments_dataset.csv")

products = spark.read.option("header", True).option("inferSchema", True) \
    .csv(base_path + "olist_products_dataset.csv")
```

```python
customers.write.mode("overwrite").saveAsTable("bronze_customers")
orders.write.mode("overwrite").saveAsTable("bronze_orders")
order_items.write.mode("overwrite").saveAsTable("bronze_order_items")
payments.write.mode("overwrite").saveAsTable("bronze_payments")
products.write.mode("overwrite").saveAsTable("bronze_products")
```

silver 

```python
orders_clean = spark.table("bronze_orders") \
    .filter(col("order_status") == "delivered") \
    .dropDuplicates()

orders_clean.write.mode("overwrite").saveAsTable("silver_orders")

customers_clean = spark.table("bronze_customers").dropDuplicates()
customers_clean.write.mode("overwrite").saveAsTable("silver_customers")

order_items_clean = spark.table("bronze_order_items").dropDuplicates()
order_items_clean.write.mode("overwrite").saveAsTable("silver_order_items")

payments_clean = spark.table("bronze_payments").dropDuplicates()
payments_clean.write.mode("overwrite").saveAsTable("silver_payments")

products_clean = spark.table("bronze_products").dropDuplicates()
products_clean.write.mode("overwrite").saveAsTable("silver_products")
```

GOLD LAYER (JOINED DATASET)

```python
gold_df = customers_clean \
    .join(orders_clean, "customer_id") \
    .join(order_items_clean, "order_id") \
    .join(payments_clean, "order_id") \
    .join(products_clean, "product_id")

gold_df.write.mode("overwrite").saveAsTable("gold_ecommerce")

display(gold_df)
```

RFM FEATURE ENGINEERING

```python
rfm_df = gold_df.groupBy("customer_id") \
    .agg(
        datediff(lit(max_date), max("order_purchase_timestamp")).alias("recency"),
        count("order_item_id").alias("frequency"),
        sum("payment_value").alias("monetary")
    )
display(rfm_df)
```

```python
max_date = gold_df.select(max("order_purchase_timestamp")).collect()[0][0]

rfm_df = gold_df.groupBy("customer_id") \
    .agg(
        datediff(lit(max_date), max("order_purchase_timestamp")).alias("recency"),
        count("order_item_id").alias("frequency"),
        sum("payment_value").alias("monetary")
    )
#display(rfm_df)
rfm_df.write.mode("overwrite").saveAsTable("gold_customer_rfm")

display(rfm_df)
```

FEATURE SCALING + K-MEANS

```python
assembler = VectorAssembler(
    inputCols=["recency", "frequency", "monetary"],
    outputCol="features"
)

feature_df = assembler.transform(rfm_df)

scaler = StandardScaler(
    inputCol="features",
    outputCol="scaled_features"
)

scaler_model = scaler.fit(feature_df)
scaled_df = scaler_model.transform(feature_df)
```

final clustering 

```python
kmeans = KMeans(featuresCol="scaled_features", k=4, seed=42)
model = kmeans.fit(scaled_df)

clustered_df = model.transform(scaled_df)

clustered_df.select(
    "customer_id", "recency", "frequency", "monetary", "prediction"
).display()
```

clutering analysis

```python
clustered_df.groupBy("prediction") \
    .agg(
        avg("recency").alias("avg_recency"),
        avg("frequency").alias("avg_frequency"),
        avg("monetary").alias("avg_monetary"),
        count("customer_id").alias("num_customers")
    ).display()
```

visualization

```python
pdf = clustered_df.select(
    "frequency", "monetary", "prediction"
).toPandas()

plt.figure(figsize=(8, 6))
plt.scatter(pdf["frequency"], pdf["monetary"], c=pdf["prediction"], cmap="viridis")
plt.xlabel("Frequency")
plt.ylabel("Monetary")
plt.title("Customer Segmentation using RFM (K-Means)")
plt.colorbar(label="Cluster")
plt.show()
```

```python
from pyspark.sql.functions import when

labeled_df = clustered_df.withColumn(
    "segment",
    when(col("prediction") == 2, "VIP High-Value Customers")
    .when(col("prediction") == 3, "New Customers")
    .when(col("prediction") == 0, "At-Risk Customers")
    .when(col("prediction") == 1, "Lost Customers")
)
```

```python
labeled_df.groupBy("segment") \
    .agg(
        avg("recency").alias("avg_recency_days"),
        avg("frequency").alias("avg_orders"),
        avg("monetary").alias("avg_spend"),
        count("customer_id").alias("customers")
    ).display()
```

```python
labeled_df.groupBy("segment").count().display()
```

```python
from pyspark.sql.functions import when

final_customer_analysis = clustered_df.withColumn(
    "customer_type",
    when(col("prediction") == 2, "VIP / High-Value Customer")
    .when(col("prediction") == 3, "New Customer")
    .when(col("prediction") == 0, "At-Risk Customer")
    .when(col("prediction") == 1, "Lost Customer")
).withColumn(
    "business_action",
    when(col("prediction") == 2, "Offer loyalty rewards and exclusive deals")
    .when(col("prediction") == 3, "Send welcome offers and onboarding emails")
    .when(col("prediction") == 0, "Send discount coupons to re-engage")
    .when(col("prediction") == 1, "Low priority; occasional win-back campaign")
)

final_customer_analysis.select(
    "customer_id",
    "recency",
    "frequency",
    "monetary",
    "customer_type",
    "business_action"
).limit(20).display()
```
